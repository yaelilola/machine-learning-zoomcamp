Q6. Median value of horsepower
Find the median value of the horsepower column in the dataset.
Next, calculate the most frequent value of the same horsepower column.
Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
Now, calculate the median value of horsepower once again.
Has it changed?

Yes, it increased
Yes, it decreased
No

A6: Yes, it decreased (because horsepower_median > horsepower_median_filled)

```python
horsepower_median = df.horsepower.describe()['50%']
most_frequent_horsepower = df.horsepower.value_counts().idxmax()
df_with_max_horsepower_for_missing = df.fillna({'horsepower': most_frequent_horsepower})
horsepower_median_filled = df_with_max_horsepower_for_missing.horsepower.describe()['50%']
horsepower_median > horsepower_median_filled
print(f"Median of horsepower before filling missing values: {horsepower_median}")
print(f"Median of horsepower after filling missing values: {horsepower_median_filled}")
print(f"horsepower_median {"decreased" if horsepower_median > horsepower_median_filled else "increased" if horsepower_median < horsepower_median_filled else "same"} after filling missing values.")
```

Q7. Sum of weights
Select all the cars from Asia
Select only columns vehicle_weight and model_year
Select the first 7 values
Get the underlying NumPy array. Let's call it X.
Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
Invert XTX.
Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
Multiply the inverse of XTX with the transpose of X, and then multiply the tresult by y. Call the result w.
What's the sum of all the elements of the result?
Note: You just implemented linear regression. We'll talk about it in the next lesson.

0.0369
0.369
3.69
36.9

A7: 0.369

```python
import numpy as np
asian_cars = df[df['origin'] == 'Asia']
X = asian_cars[['vehicle_weight', 'model_year']].head(7).to_numpy()
X
XTX = X.T.dot(X)
XTX_INVERSE = np.linalg.inv(XTX)
y =  [1100, 1300, 800, 900, 1000, 1100, 1200]
w = XTX_INVERSE.dot(X.T).dot(y)
w.sum()
```
